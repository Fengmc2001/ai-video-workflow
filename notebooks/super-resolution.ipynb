在「运行时 → 更改运行时类型」中选择 GPU。先用一段短片确认效果，再处理完整视频。此笔记本沿用原片帧率与同期声；60fps 插帧在上传前完成。

```python
import torch
assert torch.cuda.is_available(), "请先选择 GPU 运行时。"
print(torch.cuda.get_device_name(0))
```

```python
from pathlib import Path
from google.colab import files
uploaded = files.upload()
assert len(uploaded) == 1, "每次上传一段视频。"
source = Path(next(iter(uploaded)))
```

```python
import subprocess
output = Path("film-1080p.mp4")
subprocess.run(["aivideo", "upscale", str(source), "--output", str(output),
                "--device", "cuda", "--strength", "0.85"], check=True)
```

```python
files.download(str(output))
files.download(str(output.with_suffix(".report.json")))
```

下载后可在本地运行 aivideo inspect film-1080p.mp4 --fps 60 --frames --decode，核对帧率和完整解码。处理完成后，在 Colab 断开并删除运行时。
